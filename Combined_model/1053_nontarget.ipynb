```python
def calculate_percentage_change(group):
    max_lssld = group['LSSLD'].max()
    min_lssld = group['LSSLD'].min()
    if max_lssld != 0:
        group['LSSLD_PCT_CHANGE'] = 100 * (max_lssld - min_lssld) / max_lssld
    else:
        group['LSSLD_PCT_CHANGE'] = 100
    return group.iloc[[0]].drop(columns=['LSSLD'])
```

```python
result_df = x.groupby(['SUBJID', 'LSSITE']).apply(calculate_percentage_change).reset_index(drop=True)
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectFromModel
```

```python
Xtemp = df1.drop('LSSLD_Change', axis=1)
ytemp = df1['LSSLD_Change'].astype(int)

#Random Forest classifier
rf = RandomForestClassifier()
rf.fit(Xtemp, ytemp)

selector = SelectFromModel(rf, prefit=True, max_features=40)
selected_features = Xtemp.columns[selector.get_support()]

print("Top 40 features:")
print(selected_features)
```

```python
selected_features = list(selected_features) + ['LSSLD_Change'] +matching_columns
```

```python
df1 = df1[selected_features]
```

```python
X1 = df1.drop(columns=['LSSLD_Change','SUBJID'], axis=1)
y1 = df1['LSSLD_Change'].astype(int)
X_train, X_test, y_train, y_test = train_test_split(X1, y1, test_size=0.2, random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
param_grids = {
    'Logistic Regression': {
        'C': [0.001, 0.01, 0.1, 1, 10, 100],
        'penalty': ['l1', 'l2'],
        'solver': ['liblinear']  
    },
    'Random Forest': {
        'n_estimators': [50, 100, 200],
        'max_depth': [None, 10, 20],
        'min_samples_split': [2, 5],
        'min_samples_leaf': [1, 2]
    },
    'SVM': {
        'C': [0.1, 1, 10],
        'kernel': ['rbf', 'linear'],
        'gamma': ['scale', 'auto']
    },
    'XGBoost': {
        'n_estimators': [50, 100, 200],
        'max_depth': [3, 5, 7],
        'learning_rate': [0.01, 0.1, 0.3],
        'subsample': [0.8, 0.99],
        'colsample_bytree': [0.8, 0.99]
    }
}
```

```python
models = {
    'Logistic Regression': LogisticRegression(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'SVM': SVC(random_state=42),
    'XGBoost': XGBClassifier(random_state=42)
}
```

```python
for name, model in models.items():
    print(f"Tuning {name}...")
    grid_search = GridSearchCV(estimator=model, param_grid=param_grids[name], 
                               cv=5, n_jobs=-1, verbose=1, scoring='accuracy')
    grid_search.fit(X_train_scaled, y_train)
    
    print(f"Best parameters for {name}: {grid_search.best_params_}")
    print(f"Best cross-validation score: {grid_search.best_score_:.4f}")
    
    best_model = grid_search.best_estimator_
    y_pred = best_model.predict(X_test_scaled)
    
 
    print(f"Test Accuracy: {accuracy_score(y_test, y_pred):.4f}")
    print("Classification Report:")
    print(classification_report(y_test, y_pred))
    print("\n")
```

```python
name1 = 'Random Forest'
model1 = RandomForestClassifier(random_state=42)
grid_search = GridSearchCV(estimator=model1, param_grid=param_grids[name1], 
                            cv=5, n_jobs=-1, verbose=1, scoring='accuracy')
grid_search.fit(X_train_scaled, y_train)

print(f"Best parameters for {name1}: {grid_search.best_params_}")
print(f"Best cross-validation score: {grid_search.best_score_:.4f}")

best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test_scaled)


print(f"Test Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred))
print("\n")
```
